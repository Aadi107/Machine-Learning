```python
# Customer Segmentation|Clustering System 

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("smartcart_customers.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

# Feature Engineering & Creation

```python
df.columns
```

```python
# Age

df["Age"] = 2026 - df["Year_Birth"]
```

```python
df.head()
```

```python
#Customer Joining Date

df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"],dayfirst = True)

reference_date = df["Dt_Customer"].max()

df["Customer_Tenure_Days"] = (reference_date - df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
df.columns
```

```python
# Spending

df["Total_Spending"] = df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]

df.head()
```

```python
# Children

df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
# Education

df["Education"].value_counts()

df["Education"]= df["Education"].replace({ 
    "Basic" : "Undergraduate", "2n Cycle" : "Undergraduate",
    "Graduation" : "Graduate",
    "Master" : "Postgraduate","PhD" : "Postgraduate"
})
```

```python
#Marital Status

df["Marital_Status"].value_counts()

df["Living_With"]= df["Marital_Status"].replace({ 
    "Married" : "Partner", "Together" : "Partner",
    "Single" : "Alone","Divorced" : "Alone","Widow" : "Alone",
    "Absurd":"Alone","YOLO":"Alone"
})
```

```python
df["Living_With"].value_counts()
```

# Dropping Columns

```python
df.head()
```

```python
df.shape
df.columns
```

# Feature Encoding 

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()

cat_cols = ["Education","Living_With"]                                     #Categorical Features to be encoded

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
pca.explained_variance_ratio_
```

```python
# 3D Visualisation

pca = PCA(n_components = 3)

X_pca = pca.fit_transform(X_scaled)
```

```python
pca.explained_variance_ratio_
```

```python
# Plotting

fig = plt.figure(figsize = (8,6))

ax = fig.add_subplot(111,projection = "3d")

ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2])

ax.set_xlabel("PCA 1")
ax.set_ylabel("PCA 2")
ax.set_zlabel("PCA 3")
ax.set_title("3D Projections for all PCAs")
```

# Choosing K Value

```python
# Elbow Method

from sklearn.cluster import KMeans
from kneed import KneeLocator
```

```python
wcss = []

for k in range(1,11):
    kmeans=KMeans(n_clusters = k,random_state = 42)
    kmeans.fit_transform(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee = KneeLocator(range(1,11),wcss,curve = "convex",direction = "decreasing")

optimal_K = knee.elbow
```

```python
print("Best K value = ",optimal_K)
```

```python
# Plot

plt.plot(range(1,11),wcss, marker ='o')
plt.xlabel("K")
plt.xlabel("WCSS")
```

```python
# Silhouette Score

from sklearn.metrics import silhouette_score

scores = []

for k in range (2,11):
    kmeans = KMeans(n_clusters = k , random_state = 42)
    label = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca , label)
    scores.append(score)
```

```python
#Plotting 

plt.plot(range(2,11),scores, marker ='o')
plt.xlabel("K")
plt.xlabel("Silhouette Score")
```

```python
# Combined Plot (K vs WCSS)

k_range = range(2,11)

fig,ax1 = plt.subplots(figsize = (8,6))

ax1.plot(k_range,wcss[:len(k_range)],marker = "o", color ="blue")

ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()
ax2.plot(k_range,scores[:len(k_range)],marker = "x",color = "red",linestyle = "--")
ax2.set_ylabel("SS")
```

# Clustering

```python
# K-Means

kmeans = KMeans(n_clusters = 4,random_state = 42)
labels_kmeans = kmeans.fit_predict(X_pca) 
```

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection = "3d")

ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c = labels_kmeans)
```

```python
#Agglomerative Clustering

from sklearn.cluster import AgglomerativeClustering
```

```python
agg_cl = AgglomerativeClustering(n_clusters = 4, linkage = "ward")
labels_agg = agg_cl.fit_predict(X_pca)

fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection = "3d")

ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c = labels_agg)
```

##### Agglomerative Clustering is thus the best clustering here as it clearly seperates data into four clusters (DBSCAN returns 30+ clusters)
